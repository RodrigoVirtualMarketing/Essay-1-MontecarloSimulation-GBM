```python

import pandas as pd
import numpy as np

# Cargar el DataFrame desde un archivo CSV (ajustar a la ruta correcta)
Ufc_db = pd.read_csv('Final_Ufc_db.csv')

# Establecer el índice (opcional si necesitas que 'fight_id' sea el índice)
#Ufc_db.set_index('fight_id', inplace=True) # Removing this line to keep fight_id as a column

# Listas de IDs de peleadores
lista_peleadores1 = [1664, 1664, 1664]
lista_peleadores2 = [899, 899, 899]

# DataFrames vacíos para almacenar las últimas peleas de cada lista
ultimas_peleas_1 = pd.DataFrame()
ultimas_peleas_2 = pd.DataFrame()

# Obtener la última pelea de cada peleador en lista_peleadores1
for fighter_id in lista_peleadores1:
    # Filtrar las filas donde el peleador aparece como 'f1_fighter_w' o 'f2_fighter_w'
    ultima_pelea = Ufc_db[(Ufc_db['f1_fighter_w'] == fighter_id) | (Ufc_db['f2_fighter_w'] == fighter_id)].nlargest(1, 'fight_id')
    # Agregar la pelea al DataFrame de últimas peleas
    ultimas_peleas_1 = pd.concat([ultimas_peleas_1, ultima_pelea], ignore_index=True)

# Obtener la última pelea de cada peleador en lista_peleadores2
for fighter_id in lista_peleadores2:
    # Filtrar las filas donde el peleador aparece como 'f1_fighter_w' o 'f2_fighter_w'
    ultima_pelea = Ufc_db[(Ufc_db['f1_fighter_w'] == fighter_id) | (Ufc_db['f2_fighter_w'] == fighter_id)].nlargest(1, 'fight_id')
    # Agregar la pelea al DataFrame de últimas peleas
    ultimas_peleas_2 = pd.concat([ultimas_peleas_2, ultima_pelea], ignore_index=True)

# Unir las dos listas de últimas peleas en un DataFrame
emparejamientos = pd.concat([ultimas_peleas_1.reset_index(drop=True), ultimas_peleas_2.reset_index(drop=True)], axis=1)

# Mostrar el DataFrame final de emparejamientos
print(emparejamientos)

# Si deseas eliminar algunas columnas (por ejemplo, 'f1_fighter_w', 'f2_fighter_w'), puedes hacerlo así:
Ufc_db = Ufc_db.drop(['f1_fighter_w', 'f2_fighter_w'], axis=1)
```

```python

# Establecemos listas de columnas para f1 y f2 para extracción.
f1_columns = [col for col in Ufc_db.columns if 'f1_' in col]
f2_columns = [col for col in Ufc_db.columns if 'f2_' in col]

# Para f1_stats
f1_stats = pd.DataFrame()
for index, row in ultimas_peleas_1.iterrows():
    if row['f1_fighter_id'] in lista_peleadores1:
        # Convertimos la Serie a DataFrame y transponemos
        data = row[f1_columns].to_frame().T.rename(lambda x: x.replace('f1_', ''), axis='columns')
        f1_stats = pd.concat([f1_stats, data], ignore_index=True)
    else:
        data = row[f2_columns].to_frame().T.rename(lambda x: x.replace('f2_', ''), axis='columns')
        f1_stats = pd.concat([f1_stats, data], ignore_index=True)

# Repetimos el proceso para f2_stats
f2_stats = pd.DataFrame()
for index, row in ultimas_peleas_2.iterrows():
    if row['f1_fighter_id'] in lista_peleadores2:
        data = row[f1_columns].to_frame().T.rename(lambda x: x.replace('f1_', ''), axis='columns')
        f2_stats = pd.concat([f2_stats, data], ignore_index=True)
    else:
        data = row[f2_columns].to_frame().T.rename(lambda x: x.replace('f2_', ''), axis='columns')
        f2_stats = pd.concat([f2_stats, data], ignore_index=True)

matchups = pd.concat([f1_stats, f2_stats], axis=1)

# Crear la columna 'num_rounds' con los valores que has proporcionado
matchups['num_rounds'] = [5, 3, 3, 3, 3]
matchups['title_fight'] = [1, 0, 0, 0, 0]
# Verificar que la columna se haya añadido correctamente
matchups
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor    # para usar modelos de tipo arboles de decision
from sklearn.model_selection import cross_val_score   # para evaluar algoritmos en cross validacion
from sklearn.model_selection import KFold  # para partir la data en k-folds
# carga funciones de Modelos de ensamble
from sklearn.ensemble import AdaBoostRegressor   # Ensamble AdaBoost para regresion
from sklearn.ensemble import GradientBoostingRegressor   # Ensamble Gradient Boosting para regresion
from sklearn.ensemble import RandomForestRegressor   # Ensamble RandomForest para regresion
from sklearn.ensemble import ExtraTreesRegressor     # Ensamble ExtraTrees para regresion
```

```python
array = Ufc_db.values
X = array[:,0:30]  #atributos
y = array[:,0]
    #target
X
```

```python
#Normalización- No deberia ser por División
scalerE = StandardScaler().fit(X)
rescaledX = scalerE.transform(X)
#print(rescaledX)
```

```python
test_size = 0.2    # proporcion de muestras para validacion
seed = 7   # semilla para el generador aleatorio (para reproductibilidad)
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=test_size, random_state=seed)

print('Training: %d ejemplos. Validation: %d ejemplos' % (len(y_train), len(y_valid)))
```

```python
# Definimos el pool de algoritmos de ensamble a evaluar
algos = []
algos.append(('DecisionTree', DecisionTreeRegressor())) # secuencial
algos.append(('AdaBoost', AdaBoostRegressor())) # secuencial
algos.append(('GradientBoosting', GradientBoostingRegressor())) # secuencial
algos.append(('RandomForest', RandomForestRegressor())) # paralelo
algos.append(('ExtraTrees', ExtraTreesRegressor())) # paralelo
```

```python
#Evaluamos cada algoritmo de Ensamble en estrategia de 5-fold-CV
results = []
names = []

# genera el particionamiento de 5 folds que seran usados en cada evaluacion
seed = 7
kfold = KFold(n_splits=5, random_state=seed, shuffle= True)  # especifica el particionador de datos a 5-folds CV

#  evalua cada algoritmo
for algoname, algo in algos:
    cv_results = cross_val_score(algo, X_train, y_train, cv=kfold, scoring='neg_mean_squared_error')  #r2 neg_mean_squared_error
    results.append(cv_results)
    names.append(algoname)
    print("{}: {} ({})".format(algoname, cv_results.mean(), cv_results.std()))
```

```python
# Muestra graficamente los resultados de neg_mean_absolute_error  en crossvalidacion de los Algoritmos evaluados
fig = plt.figure(figsize=[10,7])
fig.suptitle('neg_mean_squared_error  obtenidas en 5-fold-CV')
ax = fig.add_subplot(111)
plt.boxplot(results)
ax.set_xticklabels(names)
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import KFold  # para partir la data en k-folds
```

```python
# define los valores que seran probados de los hiperparametros max_depth y n_estimators
parameter_grid = dict(max_depth=np.array([3,10,50]),
                      n_estimators=np.array([90,100,110]))

# define los folds para la evaluacion de cada valor del hiperparametro n_estimators
seed = 7
kfold = KFold(n_splits=5, random_state=seed, shuffle= True)  # especifica el particionador de datos a 10-folds CV

# define el tipo de modelo con el cual se testara diferentes hyperparametros
model = GradientBoostingRegressor()

# define el buscador grid en crosvalidacion
grid = GridSearchCV(estimator=model, param_grid=parameter_grid, scoring='neg_mean_squared_error', cv=kfold)
grid_result = grid.fit(X_train, y_train)

# muestra resultados de la busqueda grid
print("Mejor neg_mean_squared_error: %f con hyperparametro %s" % (grid_result.best_score_, grid_result.best_params_))
means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']
for mean, stdev, param in zip(means, stds, params):
    print("%f (%f) with: %r" % (mean, stdev, param))
```
